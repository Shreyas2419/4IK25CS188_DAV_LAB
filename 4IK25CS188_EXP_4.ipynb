```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("/content/student_performance_dataset.csv")
print("First 5 rows of dataset:")
print(df.head())
```

```python
print("\nDataset Info:")
print(df.info())
```

```python
print("Missing values before handling:")
print(df.isnull().sum())
```

```python
df.loc[0:5,'Study_Hours_per_Week']=None
df.loc[10:15,'Attendance_Rate']=None
df.loc[20:25, 'Past_Exam_Scores']=None
df.loc[30:35, 'Final_Exam_Scores']=None
print("Missing values after inserting sample NaNs:")
print(df.isnull().sum()[df.isnull().sum()>0])
```

```python
df['Study_Hours_per_Week'] = df['Study_Hours_per_Week'].fillna(df['Study_Hours_per_Week'].mean())
```

```python
df['Attendance_Rate'] = df['Attendance_Rate'].fillna(df['Attendance_Rate'].median())
```

```python
df['Past_Exam_Scores'] = df['Past_Exam_Scores'].ffill()
```

```python
df['Final_Exam_Score'] = df['Final_Exam_Score'].fillna(df['Final_Exam_Score'].mean())
print("Missing values after complete imputation:")
print(df[['Study_Hours_per_Week', 'Attendance_Rate', 'Past_Exam_Scores', 'Final_Exam_Score']].isnull().sum())
```

```python
if 'Parental_Education_Level' in df.columns:
    df = pd.get_dummies(df, columns=['Parental_Education_Level'])
else:
    print("Parental_Education_Level is already one-hot encoded or not present.")
print("\nOne-Hot Encoded columns:")
print([col for col in df.columns if 'Parental_Education_Level' in col])
```

```python
le=LabelEncoder()
df["Pass_Fail_Encoded"]= le.fit_transform(df['Pass_Fail'])
print("\nLabel Encoded Results:")
print(df[['Pass_Fail','Pass_Fail_Encoded']].head())
```

```python
le = LabelEncoder()
df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df['Internet_Access_Encoded'] = le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Encoded'] = le.fit_transform(df['Extracurricular_Activities'])
```

```python
df.to_csv('student_performance_cleaned.csv', index=False)
print("Saved the final cleaned dataset as 'student_performance_cleaned.csv'")
```
