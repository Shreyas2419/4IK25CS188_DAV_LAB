# Program 3 : Data Visualization using matplotlib and seaborn: Histogram , Box plot , Scatter plot , Pair plots & Correlation heatmaps.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Loading Data

```python
df =pd.read_csv("/content/student_data.csv")
print("Dataset shape:", df.shape)
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
print(df.describe())
```

# 1. Histogram - distribution of final grade

```python
plt.figure(figsize=(6, 4))
sns.histplot(df['G3'], bins=20, kde=True , color= "steelblue")
plt.title('Distribution of Final Grade (G3)')
plt.xlabel('Final Grade')
plt.ylabel('Frequency')
plt.show()
```

# 2. Box Plot - grade spread across study time levels

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x='studytime', y='G3', data=df, palette="Set2")
plt.title('Final Grade Spread Across Study Time Levels')
plt.xlabel('Study Time (1=low ,4=high)')
plt.ylabel('Final Grade')
plt.show()
```

# 3. Scatter Plot - Relationship between two grade periods

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(x='G1', y='G2',hue="sex" ,data=df , palette="coolwarm")
plt.title('First Grade Periods vs Final Grade')
plt.xlabel('G1 (Period 1 Grade)')
plt.ylabel('G3 (First Grade)')
plt.show()
```

# 4. Pair Plot - Pairwise relationship across key numeric columns

```python
sns.pairplot(df[["age","studytime","G1","G2","G3","sex"]],hue="sex", palette="husl")
plt.suptitle('Pairwise Relationship Across Numeric Columns', y=1.02)
plt.show()
```

# 5. Correlation Heatmap - Strength of relationships between numeric features

```python
numeric_cols = ["age","studytime","failures","G1","G2","G3"]
plt.figure(figsize=(8, 6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
